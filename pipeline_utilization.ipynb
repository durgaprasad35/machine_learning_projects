```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Importing the pipeline library 
from sklearn.pipeline import Pipeline 
from sklearn.linear_model import LogisticRegression 
```

```python
from sklearn.preprocessing import StandardScaler
steps = [('scalar',StandardScaler()) , ('classifier',LogisticRegression())] 
```

```python
pipeline = Pipeline(steps) 
```

```python
pipeline
```

```python
#Importing the dataset 
from sklearn.datasets import load_diabetes 
X,y = load_diabetes(return_X_y=True) 
```

```python
#Spliting the dataset into the training and testing parts 
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.33,random_state=42) 
```

```python
#Now fitting the data into the pipeline 
pipeline.fit(X_train,y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.impute import SimpleImputer 
from sklearn.preprocessing import OneHotEncoder
```

```python
#Creating a pipeline for the numerical columns preprocessing
steps = [('Imputer_mean',SimpleImputer(missing_values=np.nan,strategy='mean')),('scalar',StandardScaler())]

numerical_preprocessor = Pipeline(steps)
```

```python
numerical_preprocessor
```

```python
#Creating a pipeline for the categorical columns preprocessing 
steps = [('Impute_constant',SimpleImputer(fill_value='missing',strategy='constant')),('scalar',StandardScaler()),('encoding',OneHotEncoder(handle_unknown='ignore'))] 
```

```python
categorical_preprocessor = Pipeline(steps)
```

```python
categorical_preprocessor
```

```python
#now we try to combine these two preprocessing pipelines into one pipeline 
from sklearn.compose import ColumnTransformer
```

```python
preproccesor = ColumnTransformer([('categorical',categorical_preprocessor,['gender']),('numerical',numerical_preprocessor,['sales'])])
```

```python
from sklearn.pipeline import make_pipeline
```

```python
pipe = make_pipeline(preproccesor,LogisticRegression())
```
